```python
# 1. Data Exploration and Preprocessing
# ●	Begin by loading and exploring the "Alphabets_data.csv" dataset. Summarize its key features such as the number of samples, features, and classes.
# ●	Execute necessary data preprocessing steps including data normalization, managing missing values.
```

```python
# ●	Begin by loading and exploring the "Alphabets_data.csv" dataset. Summarize its key features such as the number of samples, features, and classes.
import pandas as pd
df= pd.read_csv("sonardataset.csv")
df.head()
```

```python
df.shape # There are 208 columns or features and 61 rows or records
```

```python
df['Y'].value_counts() # There are two classes M (111) and R(97)
```

```python
# ●	Execute necessary data preprocessing steps including data normalization, managing missing values.
df.isnull().sum() # There are no null values in the given dataset
```

```python
# Normalization
df_count= df.drop('Y',axis=1)
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df_scaled= scaler.fit_transform(df_count)
df_scaled = pd.DataFrame(df_scaled,columns= df_count.columns)
df_scaled
```

```python
# Applying Label encoding for Y variable
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Y'] = le.fit_transform(df['Y'])
```

```python
X= df_scaled
Y= df['Y']
```

```python
# 2. Model Implementation
# ●	Construct a basic ANN model using your chosen high-level neural network library. Ensure your model includes at least one hidden layer.
# ●	Divide the dataset into training and test sets.
# ●	Train your model on the training set and then use it to make predictions on the test set.
```

```python
# ●	Divide the dataset into training and test sets.
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.3,random_state = 42) 
X_train.shape
```

```python
Y_train.shape
```

```python
X_test.shape
```

```python
Y_test.shape
```

```python
# ●	Construct a basic ANN model using your chosen high-level neural network library. Ensure your model includes at least one hidden layer.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
model = Sequential()
model.add(Dense(60,activation = 'relu',input_shape=(60,))) # hidden layer 1
model.add(Dense(90,activation ='relu')) # hidden layer 2
model.add(Dense(1,activation = 'sigmoid')) # output layer
# compile the model
model.compile(optimizer ='Adam',loss='binary_crossentropy',metrics=['accuracy'])

# ●	Train your model on the training set and then use it to make predictions on the test set.
import numpy as np
history = model.fit(X_train,Y_train,epochs = 50,batch_size=10)
loss,accuracy = model.evaluate(X_test,Y_test,verbose = 1)
print("Test Loss for initial model:", np.round(loss,2))
print("Test Accuracy for initial model:",np.round( accuracy,2))  # Test Loss: 0.29 ,Test Accuracy: 0.89
```

```python
# predicting on the test data
Y_pred = model.predict(X_test)
Y_pred_class = (Y_pred>0.5).astype(int) 
print(f"Y_predicted first 10 classes are {Y_pred_class[:10]}")
```

```python
# ●	Adopt a structured approach like grid search or random search for hyperparameter tuning, documenting your methodology thoroughly.
import pandas as pd
import numpy as np
from scikeras.wrappers import KerasClassifier
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
```

```python
# creating model function
def create_model(hidden1=64,hidden2=32,activation1='relu',activation2='relu',learning_rate=0.01):
    model5=Sequential()
    model5.add(Dense(hidden1,activation= activation1,input_shape=(60,)))
    model5.add(Dense(hidden2,activation=activation2))
    model5.add(Dense(1,activation = "sigmoid"))

    optimizer = Adam(learning_rate = learning_rate)
    model5.compile(optimizer= optimizer,loss='binary_crossentropy',metrics=['accuracy'])
    return model5    
```

```python
# GridsearchCV
best_acc = 0
best_params = None

for h1 in [32, 64]:
    for h2 in [16, 32]:
        for act1 in ['relu','tanh']:
            for act2 in ['relu','tanh']:
                for lr in [0.001,0.01]:
                    model5 = create_model(hidden1=h1, hidden2=h2, activation1=act1, activation2=act2, learning_rate=lr)
                    model5.fit(X_train, Y_train, epochs=50, batch_size=10, verbose=1)
                    loss, acc = model5.evaluate(X_test, Y_test, verbose=1)
                    if acc > best_acc:
                        best_acc = acc
                        best_params = (h1,h2,act1,act2,lr)

print("Best Accuracy:", best_acc)
print("Best Params:", best_params)  # Best Accuracy: 0.9047619104385376, Best Params: (32, 16, 'tanh', 'relu', 0.01)
```
